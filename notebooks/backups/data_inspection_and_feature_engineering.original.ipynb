### Match table conclusion

- The Match table is the main table for this project because one row represents one football match.
- It contains the match date, season, stage, league, home team, away team, player IDs, player positions and final goals.
- `home_team_goal` and `away_team_goal` will be used to create the target variable:
  - Home Win
  - Draw

### Team attribute coverage decision

The match dataset covers 2008/2009 to 2015/2016, while the Team_Attributes table begins on 22 February 2010.

As a result:

- 2008/2009 has no historical team attribute information.
- 2009/2010 contains only partial team attribute coverage.
- Seasons from 2010/2011 onward have much higher coverage.

To avoid using future team information and causing data leakage, the modelling dataset was restricted to complete seasons from 2010/2011 to 2015/2016.

Matches where either the home or away team still had no historical Team_Attributes record were excluded rather than imputing artificial tactical values.

This produces a cleaner dataset where team attributes represent information that was genuinely available before each match.

```python
# We calculate form from ALL historical matches
history_source = match_base[
    [
        "id",
        "date",
        "home_team_api_id",
        "away_team_api_id",
        "home_team_goal",
        "away_team_goal"
    ]
].copy()

history_source["date"] = pd.to_datetime(history_source["date"])
```

```python
away_history = history_source[
    [
        "id",
        "date",
        "away_team_api_id",
        "home_team_api_id",
        "away_team_goal",
        "home_team_goal"
    ]
].copy()

away_history.columns = [
    "match_id",
    "date",
    "team_api_id",
    "opponent_api_id",
    "goals_for",
    "goals_against"
]
```

```python
team_history = pd.concat(
    [home_history, away_history],
    ignore_index=True
)

team_history.head()
```

```python
team_history.shape
```

```python
def calculate_points(row):
    if row["goals_for"] > row["goals_against"]:
        return 3
    elif row["goals_for"] == row["goals_against"]:
        return 1
    else:
        return 0


team_history["points"] = team_history.apply(
    calculate_points,
    axis=1
)
```

```python
team_history[
    [
        "team_api_id",
        "date",
        "goals_for",
        "goals_against",
        "points"
    ]
].head(10)
```

```python
team_history = team_history.sort_values(
    ["team_api_id", "date", "match_id"]
).reset_index(drop=True)
```

```python
team_history["form_points_5"] = (
    team_history
    .groupby("team_api_id")["points"]
    .transform(
        lambda x: x.shift(1).rolling(
            window=5,
            min_periods=5
        ).sum()
    )
)
```

```python
team_history["avg_goals_scored_5"] = (
    team_history
    .groupby("team_api_id")["goals_for"]
    .transform(
        lambda x: x.shift(1).rolling(
            window=5,
            min_periods=5
        ).mean()
    )
)
```

```python
team_history["avg_goals_conceded_5"] = (
    team_history
    .groupby("team_api_id")["goals_against"]
    .transform(
        lambda x: x.shift(1).rolling(
            window=5,
            min_periods=5
        ).mean()
    )
)
```

### Recent-form feature decision

Recent team performance was calculated using each team's previous five matches.

The following features were created:

- points earned in the previous five matches
- average goals scored in the previous five matches
- average goals conceded in the previous five matches

The current match was excluded from all calculations using a one-match shift to prevent data leakage.

252 matches (approximately 1.38% of the modelling dataset) did not have a complete five-match history for both teams. These matches were removed instead of imputing artificial recent-form values.

After this step, 18,016 matches remained.

### Player attribute coverage decision

Historical player attributes were matched to each player using the latest available attribute record before the match date.

Most matches had very strong player-data coverage. The median number of players with available historical attributes was 11 for both home and away teams, with an average of approximately 10.83 players per team.

A small number of matches contained very limited player information. To ensure that aggregated team-level player features were based on a reasonable portion of the lineup, matches were required to have historical attributes for at least 8 players on both teams.

Matches below this threshold were removed rather than estimating large portions of the lineup.

Player features were aggregated into team-level measures such as overall rating, attacking ability, passing, technical ability, pace, physical ability, defence and goalkeeper strength.
